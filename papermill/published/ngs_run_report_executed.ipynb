```python
from pathlib import Path
import os
import pandas as pd

print("run_id:", run_id)
print("input_root:", input_root)
print("notes:", notes or "(none)")
root = Path(input_root)
if not root.exists():
    print("input_root path does not exist yet — create it or set params to your data directory.")
else:
    csvs = sorted(root.rglob("*.csv"))
    print(f"Found {len(csvs)} CSV files under {root.resolve()}")
    for p in csvs[:5]:
        print(" -", p.relative_to(root))
    if csvs:
        sample = csvs[0]
        try:
            df = pd.read_csv(sample, nrows=5)
            print("Sample columns:", list(df.columns))
        except Exception as exc:
            print("Could not preview", sample, ":", exc)
```
